## 1. Investigación Preliminar

**¿Qué es un filtro y por qué es importante?**  
Leyendo el material y buscando información, entendí que un filtro digital no es solo un "efecto visual" como los que usamos en las redes sociales, sino que por detrás es un algoritmo matemático que altera el valor numérico de los píxeles. Esto es súper importante en nuestra carrera porque es el paso previo para que una computadora pueda analizar una imagen (por ejemplo, aplicando un filtro que resalte los bordes de un objeto para poder detectarlo).

**Photoshop vs. GIMP**
* **Photoshop:** Fue creado por los hermanos Knoll en 1987 y luego comprado y lanzado por Adobe en 1990. Empezó siendo algo muy básico para retocar fotos en blanco y negro, pero con los años sumó las capas (un cambio gigante) y, más recientemente, los Filtros Neuronales y el Relleno Generativo usando inteligencia artificial (Adobe Sensei). Hoy marca el ritmo de la industria.
* **GIMP:** Surgió unos años después, en 1995, como un proyecto de dos estudiantes universitarios. Su mayor diferencia es que nació siendo de código abierto (*open source*) y gratuito. Al permitir el uso de *plugins* desde casi el principio, la misma comunidad de usuarios empezó a crear y compartir sus propios filtros.

## Probando un filtro mediante código

Aprovechando que estamos usando un entorno de Notebook, me dio curiosidad ver cómo se aplica un filtro directamente sobre los píxeles usando Python, ya que en el fondo es lo mismo que hace Photoshop cuando tocamos un botón. Usé un filtro de desenfoque gaussiano muy simple con la librería `skimage`.

```python
import matplotlib.pyplot as plt
from skimage import data, filters

# 1. Cargo una imagen de prueba que ya viene en la librería
imagen_original = data.camera()

# 2. Le aplico un filtro matemático (Desenfoque Gaussiano)
imagen_filtrada = filters.gaussian(imagen_original, sigma=4)

# 3. Muestro el antes y el después para comparar
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(imagen_original, cmap='gray')
ax[0].set_title('Imagen Original', fontweight='bold')
ax[0].axis('off')

ax[1].imshow(imagen_filtrada, cmap='gray')
ax[1].set_title('Filtro de Desenfoque Aplicado', fontweight='bold')
ax[1].axis('off')

plt.tight_layout()
plt.show()
```

## 4. Actividad de Cierre: Imágenes en el Espacio

Para esta parte, le pedí a ChatGPT que me explicara cómo hacen los satélites para sacar fotos en el espacio y mandarlas a la Tierra.

**Resumen de lo que descubrí:**
* **No usan cámaras normales:** En vez de sacar fotos cuadradas como un celular, usan un sistema llamado *escáner push-broom*, que va escaneando la Tierra por franjas a medida que el satélite avanza en órbita.
* **Los filtros son cristales físicos:** Acá la palabra "filtro" se usa de forma literal. Ponen vidrios o materiales físicos frente al sensor para que solo pase un tipo de luz (infrarrojo, térmico, luz visible). Esto ayuda a captar datos invisibles al ojo humano, como la salud de un campo agrícola.
* **Transmisión de datos:** Las imágenes pesan gigabytes. Antes de enviarlas, la computadora del satélite las comprime. Actualmente, se está empezando a usar inteligencia artificial a bordo del satélite para analizar las fotos arriba mismo y borrar las que salieron tapadas por nubes; así no gastan energía enviando fotos inservibles a las antenas en la Tierra.
